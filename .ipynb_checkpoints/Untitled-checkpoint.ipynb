```python
import numpy as np
from gtsimulation.MagneticFields import magnetic_field
from gtsimulation.Global import Units


class static_dipole(magnetic_field.AbsBfield):
    def __init__(self, m_vector) -> None:
        self.m_vector = np.array(m_vector, dtype=float)


    def CalcBfield(self, x, y, z):
        mu0_4pi = 1e-7
        r_vector = np.array([x, y, z], dtype=float)
        
        r_mag = np.linalg.norm(r_vector) + 1e-11
        
            
        m_dot_r = np.dot(self.m_vector, r_vector)
        
        term1 = 3 * m_dot_r * r_vector / (r_mag ** 5)
        term2 = self.m_vector / (r_mag ** 3)
        
        B_vector = mu0_4pi * (term1 - term2)
    
        return B_vector

    def UpdateState(self, new_date):
        pass

    def to_string(self):
        return "suck`n cock"

m_vector = np.array([0, 0, 0.5e30])
T_min = 1 * Units.GeV
T_max = 1 * Units.TeV

R_star = 10000 #10km [m]
latitude = 45 / 180 * np.pi
start_posotion = np.array([np.cos(latitude), 0, np.sin(latitude)]) * R_star
```
